```python
#continue语句用于跳过当前循环的剩余语句，并继续进行下一轮循环。
# break语句用于终止当前循环，并跳出循环体。
while True:
    num = int(input("请输入一个整数（输入0退出）："))
    if num == 0:
        print("程序结束")
        break#循环终止
    elif num % 2 == 0:
        print(f"{num}是偶数")
        continue#跳过当前循环的剩余语句，继续下一轮循环        
    else:
        print(f"{num}是奇数")
```

```python
class Person():#定义一个类Person，默认继承object类
    def __init__(self,name,age,height):
        self.name=name
        self.age=age
        self.height=height
    def fun(self):
        print(f'我叫{self.name},我今年{self.age}岁,我身高{self.height}米')

class bask_person(Person):#定义一个类bask_person,继承自类Person
    def __init__(self,name,age,height):
        super().__init__(name,age,height)#调用父类的构造方法,传入name,age,height参数
        print(f'我是{self.name},我今年{self.age}岁,我身高{self.height}米,我是zmjjkk')

class sing_person(Person):#定义一个类sing_person,继承自类Person
    def __init__(self,name,age,height):
        super().__init__(name,age,height)#调用父类的构造方法,传入name,age,height参数
        print(f'我是{self.name},我今年{self.age}岁,我身高{self.height}米,我是simon')

per6 = bask_person('zmjjkk',25,1.8)#创建一个对象per6,调用bask_person类的构造方法,传入name,age,height参数
per7 = sing_person('simon',25,1.8)#创建一个对象per7,调用sing_person类的构造方法,传入name,age,height参数
per6.fun()#调用对象per6的fun方法,输出'我叫zmjjkk,我今年25岁,我身高1.8米'
per7.fun()#调用对象per7的fun方法,输出'我叫simon,我今年25岁,我身高1.8米'
```

```python
#多继承
class A():
    def printA(self):
        print('AAAAAA')

class B():
    def printB(self):
        print('BBBBBB')

class C(A,B):
    def printC(self):
        print('CCCCCC')

c1=C()#创建一个对象c1,调用C类的构造方法
c1.printA()#调用对象c1的printA方法,输出'AAAAAA'
c1.printB()#调用对象c1的printB方法,输出'BBBBBB'
c1.printC()#调用对象c1的printC方法,输出'CCCCCC'
```

```python
#多态：一个接口，多种实现，根据不同的对象，调用不同的方法
class Animal():
    def eat(self):
        pass
class Dog(Animal):#定义一个类Dog,继承自类Animal
    def eat(self):
        print('狗吃狗粮')
class Cat(Animal):#定义一个类Cat,继承自类Animal
    def eat(self):
        print('猫吃猫粮')
class Rabbit(Animal):#定义一个类Rabbit,继承自类Animal
    def eat(self):
        print('兔子吃草')

dog=Dog()
cat=Cat()
rabbit=Rabbit()

def Eat(animal):
    animal.eat()

Eat(dog)#调用函数Eat,传入对象dog,调用对象的eat方法,输出'狗吃狗粮'
Eat(cat)#调用函数Eat,传入对象cat,调用对象的eat方法,输出'猫吃猫粮'
Eat(rabbit)#调用函数Eat,传入对象rabbit,调用对象的eat方法,输出'兔子吃草'
```

```python
#魔法方法：__init__(),__str__(),__add__(),__len__(),__call__(),__getattr__(),__setattr__(),__delattr__(),__new__(),__del__(),__delattr__(),__setitem__(),__getitem__(),__contains__(),__iter__(),__next__(),__reversed__(),__reversed__(),__len__(),__str__(),__repr__(),__call__(),__call
#__init__():构造方法,用于初始化对象的属性
#__str__():将对象转换为字符串,用于打印对象
#__add__():将两个对象相加,用于拼接对象
#__len__():返回对象的长度,用于计算对象的元素数量
#__call__():将对象作为函数调用,用于调用对象的方法
#__getattr__():获取对象的属性,用于访问对象的属性
#__setattr__():设置对象的属性,用于设置对象的属性
#__delattr__():删除对象的属性,用于删除对象的属性
#__new__():创建对象,用于创建对象
#__del__():删除对象,用于删除对象
#__delattr__():删除对象的属性,用于删除对象的属性
class List_1():
    def __init__(self,lst):
        self.lst=lst
    def __getitem__(self,index):
        return self.lst[index]

my_list = [1,2,3,4,5]
wrapper=List_1(my_list)
print(wrapper.__getitem__(2))#调用对象wrapper的__getitem__方法,传入参数2,返回对象的第3个元素,输出3


```
