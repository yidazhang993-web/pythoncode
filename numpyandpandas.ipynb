```python
import numpy as np
import pandas as pd
array = np.array([[1,2,3],
                [4,5,6]])
print(array)
print('number of dim:',array.ndim)#查看数组的维度
print('shape:',array.shape)#查看数组的形状
print('size:',array.size)#查看数组的元素个数
```

```python
#创建array
a = np.array([2,23,4],dtype=np.int64)#int64,int32,float32,float64
print(a)#没有逗号，跟列表的区别
print(a.dtype)#查看数组的元素类型
b = np.array([[2,23,4]
            ,[23,11,2]])
print(b)
```

```python
a = np.zeros((3,4))#创建一个3*4的数组，元素全为0
b = np.ones((3,4))#创建一个3*4的数组，元素全为1
print(a)
print(b)
c = np.arange(12).reshape(3,4)#将一个1维数组转换为一个3*4的数组
print(c)
print(np.linspace(1,12,6).reshape(2,3))#将1到12之间的6个等间距的数转换为一个2*3的数组
```

```python
#numpy基础运算
a=np.array([10,20,30,40])
b=np.arange(4)
print(a,b)
print(a-b,a+b)
print(10*np.sin(a))#计算数组a的正弦值，结果是一个数组
print(b<3)#判断数组b中每个元素是否小于3，结果是一个布尔数组
```

```python
a=np.array([[1,1]
            ,[0,1]])
b=np.arange(4).reshape(2,2)
print(a)
print(b)
print(a*b)#数组的乘法，对应元素相乘
print(np.dot(a,b))#矩阵乘法，计算a的转置矩阵与b的乘积
print(a.dot(b))#矩阵乘法，计算a的转置矩阵与b的乘积
```

```python
a=np.random.random((3,4))#生成一个3*4的数组，元素为0到1之间的随机数
print(a)
print(np.sum(a))#计算数组a中所有元素的和
print(np.sum(a,axis=0))#计算数组a中每个列的和,axis=0表示按列进行求和
print(np.sum(a,axis=1))#计算数组a中每个行的和,axis=1表示按行进行求和
print(np.max(a))#计算数组a中所有元素的最大值
print(np.min(a))#计算数组a中所有元素的最小值
```

```python
a=np.arange(14,2,-1).reshape(3,4)
print(a)
print(np.min(a))#计算数组a中所有元素的最小值
print(np.argmin(a))#计算数组a中最小值的索引
print(np.mean(a),np.average(a),np.median(a))#计算数组a中所有元素的平均值,平均数，中位数
print(np.nonzero(a))#计算数组a中非零元素的索引
print(np.sort(a))#对数组a进行逐行排序
print(a.T,a.T.shape,a.dot(a.T))#对数组a进行转置，将行和列交换,并计算a的转置矩阵与a的乘积
print(np.clip(a,4,10))#对数组a进行裁剪，将所有元素限制在0到10之间,超出范围的元素被替换为边界值
```

```python
#numpy的索引
a=np.arange(3,15).reshape(3,4)
print(a)
print(a[2][1],a[2,1])#访问数组a的第三行的第二个列
print(a[:,1])#访问数组a的第二个列.','左右是行和列
print(a[1,:])#访问数组a的第二个行
print(a[1,1:3])#访问数组a的第二个行的第二个到第四个元素
print(a[1:3,1:3])#访问数组a的第二个到第四个行的第二个到第四个列
for row in a:#遍历数组a的行,并打印每个行的元素
    print(row)
for column in a.T:#遍历数组a的列,并打印每个列的元素
    print(column)
print(a.flatten())#将数组a展平为一维数组
for element in a.flatten():#遍历数组a的每个元素,并打印每个元素
    print(element,end="\t")
```

```python
#array的合并
a=np.array([1,1,1])
b=np.array([2,2,2])
print(np.vstack((a,b)),np.shape(np.vstack((a,b))))#垂直堆叠数组a和b,vertical stack
print(np.hstack((a,b)),np.shape(np.hstack((a,b))))#水平堆叠数组a和b,horizontal stack

c=np.array([1,1,1])[:,np.newaxis]#将数组a转换为列向量
d=np.array([2,2,2])[:,np.newaxis]#将数组b转换为列向量
print(np.hstack((c,d)),np.shape(np.hstack((c,d))))#水平堆叠数组c和d,horizontal stack

e=np.concatenate((c,d,d,c),axis=1)#水平堆叠数组c,d,d,c,horizontal stack
print(e)
```

```python
#numpy的分割
a=np.arange(12).reshape(3,4)
print(a)
print(np.split(a,2,axis=1))#按列分割
print(np.split(a,3,axis=0))#按行分割
print(np.vsplit(a,3))#按行分割
print(np.hsplit(a,2))#按列分割
```

```python
#pandas是一个基于numpy的库，用于数据处理和分析
s = pd.Series([1,3,6,np.nan,44,1])#创建一个Series对象
print(s)
```

```python
dates = pd.date_range('20260101',periods=6)#创建一个日期范围对象
print(dates)
#创建一个DataFrame对象,并指定索引和列名,数据为随机数,形状为6*4,列名分别为A,B,C,D
df = pd.DataFrame(np.random.randn(6,4),index=dates,columns=['A','B','C','D'])
print(df)
df1=pd.DataFrame(np.arange(12).reshape(4,3))
print(df1)
df2=pd.DataFrame({'A':1.0, #创建一个DataFrame对象,并指定索引和列名,数据为1.0,形状为4*1,列名分别为A
                  'B':pd.Timestamp('20260101'),
                  'C':pd.Series(1,index=list(range(4)),dtype='float32'),
                  'D':np.array([3]*4,dtype='int64'),
                  'E':pd.Categorical(['a','b','test','train']),
                  'F':'foo'})
print(df2)
print(df2.dtypes,df2.index,df2.columns)#打印DataFrame对象的索引和列名
print('*'*20)
print(df2.values)#打印DataFrame对象的值
print('*'*20)
print(df2.describe())#打印DataFrame对象的统计信息
print('*'*20)
```

```python
print(df2.T)#打印DataFrame对象的转置
print('*'*20)
print(df2.sort_index(axis=1,ascending=False))#按列索引降序排序
print('*'*20)
print(df2.sort_values(by='E'))#按列E的值升序排序
```

```python
dates = pd.date_range('20260101',periods=6)#创建一个日期范围对象,从20260101开始,共6个日期
df = pd.DataFrame(np.arange(24).reshape(6,4),index=dates,columns=['A','B','C','D'])
print(df)
print('*'*20)
print(df['A'],df.A)#打印列A的值
```

```python
print(df[0:3],df['20260101':])#打印前3行和20260101的行
print('*'*20)
print(df.loc['20260101'])#打印20260101的行，按照标签索引顺序
print('*'*20)
print(df.loc[:,['A','B']])#打印列A和B的值
```

```python
print(df.iloc[0:3,0:2])#打印前3行和前2列的值,按照索引顺序
print('*'*20)
print(df.iloc[1:3,[0,1]])#打印第2行和第3行的第1列和第2列的值,按照索引顺序
```

```python
#pandas数值设置
df.iloc[2,2]=1111   #将第3行第3列的值设为1111
df.loc['20260101','C']=2222 #将20260101的C列值设为2222
df.A[df.A>4]=3333   #将A列中大于4的值设为3333
df['F']=np.nan  #添加一列F,并用NaN填充
df['E']=pd.Series([1,2,3,4,5,6],index=df.index)  #添加一列E,并用Series填充,索引为df的索引
print(df)
print('*'*20)
```
